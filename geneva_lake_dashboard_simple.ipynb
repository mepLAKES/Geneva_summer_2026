# Why has Lake Geneva suddenly turned green?

You may have noticed that the lake's water turned turbid and green in a matter of days. Satellite images confirmed that, while low on August 25, the amount of algae had doubled by early September.

What happened, and is this unusual?

```python
from pathlib import Path
from IPython.display import Image, Video, display

PROJECT_DIR = Path.cwd()
if not (PROJECT_DIR / 'data').is_dir():
    PROJECT_DIR = PROJECT_DIR.parent

def asset(relative_path):
    path = PROJECT_DIR / relative_path
    assert path.is_file(), f'Missing dashboard asset: {path}'
    return path
```

## What the water column reveals

The CTD profiles show how temperature, chlorophyll-a, and conductivity vary with depth during the event.

```python
display(Image(filename=str(asset('graphs/ctd_temperature_chlorophyll_profiles_from_2026_08_20.png'))))
```

**CTD depth profiles after 20 August 2026:** Water temperature, chlorophyll-a, and conductivity normalized to 25 degrees C; profiles are colored by date.

**Commentary:** The wind event tilted and deepened the thermocline and brought colder water to the surface. Deep cold water is rich in nutrients such as phosphorus. While surface waters were limited in nutrients, a fresh supply of phosphorus and sunny conditions created favorable conditions for algal growth.

## Is this unusual?

The bloom is sudden, but the climatology of remote-sensed chlorophyll for Lake Geneva over the last 11 years shows that this renewed algal growth is typical for late summer. The satellite record places the 2026 chlorophyll-a observations within the seasonal distribution and year-to-year spread.

```python
context_graphs = [
    'graphs/geneva_chlorophyll_a_climatology.png',
    'graphs/geneva_air_temperature_climatology_netcdf.png',
    'graphs/geneva_wind_climatology_netcdf.png',
    'graphs/geneva_daily_mean_chlorophyll_a_violin_by_year_doy150_240.png',
]
for graph in context_graphs:
    display(Image(filename=str(asset(graph))))
```

**Context graphs:** Chlorophyll-a, air temperature, and wind climatologies with 2026 overlaid, plus annual distributions of daily mean chlorophyll-a for days of year 150-240.

**Commentary:** The wind regime for 2026 has been within typically observed values since 2019. The summer of 2026 has been 4-5 degrees C warmer than any summer observed from LéXPLORE since 2019. In spite of this unusually hot summer, the amount of algae in the lake has been consistent with values observed in previous years. Wind-driven upwellings and seiches are transient but important and recurrent mechanisms that restore nutrients to the surface waters of Lake Geneva in summer.
